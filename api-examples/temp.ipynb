```python
import json
import requests
import ipywidgets as widgets
from IPython.display import display

# Define widgets
room_id_input = widgets.Text(
    value='',
    placeholder='Enter Room ID',
    description='Room ID:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='600px')
)

file_path_input = widgets.Text(
    value='../ui/public/json_example_1.json',
    placeholder='Path to JSON file',
    description='JSON File Path:',
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='600px')
)

convert_checkbox = widgets.Checkbox(
    value=False,
    description='Convert from ASKCOS format',
    style={'description_width': 'initial'}
)

submit_button = widgets.Button(
    description='Upload JSON',
    button_style='success'
)

output = widgets.Output()

# Upload logic
def upload_json(b):
    with output:
        output.clear_output()
        room_id = room_id_input.value.strip()
        file_path = file_path_input.value.strip()
        convert = convert_checkbox.value

        if not room_id or not file_path:
            print("❌ Please provide both Room ID and file path.")
            return

        url = "http://0.0.0.0:5099/upload_json_to_ui/"

        try:
            with open(file_path, "rb") as f:
                files = {"file": ("json_example.json", f, "application/json")}
                data = {
                    "room_id": room_id,
                    "convert_from_askcos": str(convert).lower()
                }

                response = requests.post(url, data=data, files=files)

                if response.status_code == 200:
                    print("✅ Upload successful!")
                    print(json.dumps(response.json(), indent=2))
                else:
                    print("❌ Upload failed!")
                    print("Status code:", response.status_code)
                    print("Response:", response.text)

        except FileNotFoundError:
            print(f"❌ File not found: {file_path}")
        except Exception as e:
            print(f"❌ Unexpected error: {e}")

# Wire button click to function
submit_button.on_click(upload_json)

# Display the UI
display(widgets.VBox([
    room_id_input,
    file_path_input,
    convert_checkbox,
    submit_button,
    output
]))
```
